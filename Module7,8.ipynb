#1. Implement following inheritance: (1)Single (2) Multiple (3) Multilevel (4) Hybrid

```python
class Vechile:
  def __init__(self , name, color , price):
    self.name = name
    self.color = color
    self.price = price
  def info(self):
    print(self.name , self.color , self.price)

class Car(Vechile):
  def change_gear(self,no):
    print(self.name , 'change gear to number :', no)
Car1 = Car('BMWX1', 'Black' , 10000000)
Car1.info()
Car1.change_gear(2)
```

```python

class Grandfather:
  def __init__(self, name):
    self.name = name
  def talk(self):
    print(f'{self.name} says hello.')
class Father(Grandfather):
  def __init__(self, name, occupation):
    super().__init__(name)
    self.occupation = occupation
  def work(self):
    print(f'{self.name} works as a {self.occupation}.')

class Son(Father):
  def __init__(self, name, occupation, hobby):
    super().__init__(name, occupation)
    self.hobby = hobby
  def play(self):
    print(f'{self.name} enjoys {self.hobby}.')

son1 = Son('John', 'Engineer', 'coding')
son1.talk()
son1.work()
son1.play()


class Father2:
  def __init__(self, name):
    self.name = name
  def teach_math(self):
    print(f'{self.name} teaches math.')

class Mother:
  def __init__(self, name):
    self.name = name
  def teach_art(self):
    print(f'{self.name} teaches art.')

class Child(Father2, Mother):
  def __init__(self, name_child, name_father, name_mother):
    Father2.__init__(self, name_father)
    Mother.__init__(self, name_mother)
    self.name_child = name_child
    self.name_father = name_father
    self.name_mother = name_mother

  def introduce(self):
    print(f'Hi, I am {self.name_child}. My father is {self.name_father} and my mother is {self.name_mother}.')

child1 = Child('Lily', 'David', 'Sarah')
child1.introduce()
child1.teach_math()
child1.teach_art()


class Animal:
  def speak(self):
    print('Animal speaks')
class Mammal(Animal):
  def feed_milk(self):
    print('Mammal feeds milk')
class Bird(Animal):
  def lay_eggs(self):
    print('Bird lays eggs')
class Bat(Mammal, Bird):
  def fly(self):
    print('Bat can fly')


bat1 = Bat()
bat1.speak()
bat1.feed_milk()
bat1.lay_eggs()
bat1.fly()
```

#2. Demonstrate Overriding and methods to overcome.

```python
# 1. Demonstrating Method Overriding
class Vehicle:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show_details(self):
        print("Vehicle Name:", self.name, "Price:", self.price)

class Car(Vehicle):
    # Child class overrides the parent class method
    def show_details(self):
        print("Car Name:", self.name, "Special Price:", self.price)

print("--- Method Overriding ---")
c1 = Car("BMW", 10000000)
c1.show_details()  # Executes Car's show_details, overriding Vehicle's method


# 2. Methods to Overcome Method Overriding

# Way 1: Using super() function
class ElectricCar(Vehicle):
    def show_details(self):
        super().show_details()  # Calls Vehicle's show_details method
        print("Battery Type: Lithium-ion")

print("\n--- Overcoming using super() ---")
e1 = ElectricCar("Tesla", 8000000)
e1.show_details()


# Way 2: Calling Parent Class Name directly
class SportsCar(Vehicle):
    def show_details(self):
        Vehicle.show_details(self)  # Explicitly passing 'self' to parent method
        print("Top Speed: 300 km/h")

print("\n--- Overcoming using Parent Class Name ---")
s1 = SportsCar("Ferrari", 30000000)
s1.show_details()
```

#. Implement following searching techniques:(1) Linear (2) Binary

```python
# 1. Linear Search
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i  # Found, return index
    return -1  # Not found

# Driver code
numbers = [12, 45, 23, 67, 34, 89]
key = 34

result = linear_search(numbers, key)

if result != -1:
    print("Linear Search: Element found at index", result)
else:
    print("Linear Search: Element not found")


# 2. Binary Search (List must be sorted)
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid  # Found, return index
        elif arr[mid] < target:
            low = mid + 1  # Target is in right half
        else:
            high = mid - 1  # Target is in left half

    return -1  # Not found

# Driver code
sorted_numbers = [10, 20, 30, 40, 50, 60, 70]
key = 50

result = binary_search(sorted_numbers, key)

if result != -1:
    print("Binary Search: Element found at index", result)
else:
    print("Binary Search: Element not found")
```

#Implement following sorting algorithms. (1) Selection (2) Merge (3) Tim

```python
# 1. Selection Sort
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        # Swap the found minimum element with the first element
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

# Driver Code
numbers1 = [64, 25, 12, 22, 11]
selection_sort(numbers1)
print("Selection Sort result:", numbers1)


# 2. Merge Sort
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        # Recursive call on each half
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Copy data to temp lists left[] and right[]
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Checking if any element was left
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

# Driver Code
numbers2 = [38, 27, 43, 3, 9, 82, 10]
merge_sort(numbers2)
print("Merge Sort result:", numbers2)


# 3. Timsort
RUN = 32

# Helper function to perform Insertion Sort on small runs
def insertion_sort(arr, left, right):
    for i in range(left + 1, right + 1):
        temp = arr[i]
        j = i - 1
        while j >= left and arr[j] > temp:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = temp

# Helper function to merge sorted runs
def merge(arr, l, m, r):
    len1, len2 = m - l + 1, r - m
    left = arr[l : m + 1]
    right = arr[m + 1 : r + 1]

    i, j, k = 0, 0, l

    while i < len1 and j < len2:
        if left[i] <= right[j]:
            arr[k] = left[i]
            i += 1
        else:
            arr[k] = right[j]
            j += 1
        k += 1

    while i < len1:
        arr[k] = left[i]
        i += 1
        k += 1

    while j < len2:
        arr[k] = right[j]
        j += 1
        k += 1


# Driver Code
numbers3 = [5, 21, 7, 23, 19, 1, 14, 12, 40, 2, 8]
sorted(numbers3, key=None, reverse=False)
print("Timsort result:", numbers3)
```

#Implement a Singly Linked List. (Insert, Display, Delete)


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None
    def insert(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
    def delete(self, key):
        temp = self.head
        if temp is None:
            print("List is empty, cannot delete.")
            return
        if temp.data == key:
            self.head = temp.next
            temp = None
            print(f"Deleted {key}")
            return

        # Case 3: Search for the key to be deleted
        prev = None
        while temp is not None and temp.data != key:
            prev = temp
            temp = temp.next
        if temp is None:
            print(f"Value {key} not found in the list.")
            return
        prev.next = temp.next
        temp = None
        print(f"Deleted {key}")

sll = SinglyLinkedList()
sll.insert(10)
sll.insert(20)
sll.insert(30)
sll.insert(40)

print("Linked List after insertions:")
sll.display()
sll.delete(20)
sll.display()
sll.delete(10)
sll.display()
sll.delete(100)
```
