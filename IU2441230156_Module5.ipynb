#1. A program to count the number of words, number of lines, occurrence of particular word, occurrence of particular character, number of blank spaces in a text file.

```python
# Create a sample text file to test our program
sample_text = """
hello I am Samarth!

"""

with open("sample.txt", "w") as f:
    f.write(sample_text)

print("File 'sample.txt' created successfully!")
```

```python
import os

def analyze_text_file():
  filename = "sample.txt"
  target_word = input("Enter the word to search for: ").strip()
  target_char = input("Enter the character to search for: ").strip()

  if not os.path.exists("sample.txt"):
      print(f"\nError:")
      return
  line_count = 0
  word_count = 0
  blank_space_count = 0
  word_occurance = 0
  char_occurance = 0
  with open("sample.txt" , 'r') as file:
    for line in file:
      line_count+=1
      words_in_line = line.split()
      word_count += len(words_in_line)
      for words in words_in_line:
        if words == target_word:
          word_occurance+=1
      blank_space_count+=line.count(' ')
      char_occurance+=line.lower().count(target_char.lower())
    print("          FILE ANALYSIS          ")
    print(f"Total Lines               : {line_count}")
    print(f"Total Words               : {word_count}")
    print(f"Total Blank Spaces        : {blank_space_count}")
    print(f"Occurrence of '{target_word}'     : {word_occurance}")
    if len(target_char) == 1:
        print(f"Occurrence of character '{target_char}': {char_occurance}")
    else:
      print("Character search skipped (please enter exactly 1 character next time).")
      print("="*35)

# Run the analyzer
analyze_text_file()
```

#2. A program to read a string from the user and append it into a file.

```python
import os


def append_string_to_file():
  filename = "sample.txt"

  if not os.path.exists(filename):
    print(f"'{filename}' not found. A new file will be created.")
  user_string = input("Enter the string you want to append: ")
  with open(filename, "a") as file:
    file.write(user_string + "\n")
  print("Success")
append_string_to_file()
```

#3. A program to copy the contents of one file into another.

```python
import os


def copy_file_contents():
  source_file = "source.txt"
  destination_file = "destination.txt"
  if not os.path.exists(source_file):
    sample_content = (
        "Hello, this is the source file.\n"
        "We are copying this content to another file.\n"
        "Python file handling is easy!"
    )
    with open(source_file, "w") as f:
      f.write(sample_content)
    print(f"'{source_file}' created successfully with sample data!\n")
  if not os.path.exists(source_file):
    print(f"\nError: Source file '{source_file}' not found.")
    return
  with open(source_file, "r") as src:
    content = src.read()

  with open(destination_file, "w") as dest:
    dest.write(content)
print("Success")
copy_file_contents()
```

#4. A program to read a text file and print all the numbers present in the text file.

```python
import os
import re


def extract_numbers_from_file():
  filename = "sample_numbers.txt"
  sample_content = (
      "Hello Samarth, your roll number is 45 and score is 98.5.\n"
      "You have completed 3 assignments out of 5.\n"
      "The temperature dropped to -5 degrees today."
  )
  with open(filename, "w") as f:
    f.write(sample_content)
  print(f"Sample file '{filename}' created successfully!\n")
  if not os.path.exists(filename):
    print(f"\nError: File '{filename}' not found.")
    return

  numbers_found = []
  number_pattern = re.compile(r"[-+]?\d*\.\d+|\d+")

  with open(filename, "r") as file:
    for line in file:
      # Find all matches of the pattern in the current line
      found_in_line = number_pattern.findall(line)
      # Convert them to float or int depending on your preference (here kept as strings/numbers)
      for num in found_in_line:
        # Convert to float if it contains a dot, otherwise int
        parsed_num = float(num) if "." in num else int(num)
        numbers_found.append(parsed_num)
  print(f"Extracted Numbers List    : {numbers_found}")

extract_numbers_from_file()
```

#5. A program to append the contents of one file to another file.

```python
import os


def append_file_contents():
  source_file = "source.txt"
  destination_file = "destination.txt"
  sample_source_content = (
      "This is line 3 from the source file.\n"
      "This is line 4 from the source file.\n"
  )
  sample_dest_content = (
      "This is line 1 in the destination file.\n"
      "This is line 2 in the destination file.\n"
  )

  with open(source_file, "w") as f:
    f.write(sample_source_content)

  with open(destination_file, "w") as f:
    f.write(sample_dest_content)

  print("Sample files created successfully!\n")
  if not os.path.exists(source_file):
    print(f"\nError: Source file '{source_file}' not found.")
    return
  with open(source_file, "r") as src:
    source_content = src.read()
  with open(destination_file, "a") as dest:
    dest.write(source_content)
print("Success")

append_file_contents()
```

#6. A program to read a file and capitalize the first letter of every word in the file.

```python
import os


def capitalize_words_in_file():
  filename = "sample_text.txt"
  sample_content = (
      "hello world, welcome to python file handling.\n"
      "this program capitalizes the first letter of every word.\n"
      "lets see how it works in google colab!"
  )
  with open(filename, "w") as f:
    f.write(sample_content)
  print(f"Sample file '{filename}' created successfully!\n")

  if not os.path.exists(filename):
    print(f"\nError: File '{filename}' not found.")
    return
  with open(filename, "r") as file:
    original_content = file.read()
  capitalized_content = original_content.title()
  with open(filename, "w") as file:
    file.write(capitalized_content)

  with open(filename, "r") as file:
    print(file.read())

capitalize_words_in_file()
```

#7. A program to read the contents of a file in reverse order.

```python
import os


def read_file_in_reverse():
  filename = "sample_reverse.txt"
  sample_content = (
      "Line 1: Welcome to Google Colab.\n"
      "Line 2: Python file handling is powerful.\n"
      "Line 3: Reading files in reverse order.\n"
  )
  with open(filename, "w") as f:
    f.write(sample_content)
  print(f"Sample file '{filename}' created successfully!\n")

  with open(filename, "r") as file:
    lines = file.readlines()
  reversed_lines = lines[::-1]
  for line in reversed_lines:
    print(line.strip())


read_file_in_reverse()
```
