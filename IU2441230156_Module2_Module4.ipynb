2 Write a function to find out the factorial of a given number. I) without recursion II) with recursion

```python
def factorial_iteration(x):
  result = 1
  for i in range(1,x+1):
    result*=i
  return result
def factorial_recursion(x):
  if(x==0 or x==1):
    return 1
  return x*factorial_recursion(x-1)
print("the factioral in iteration",factorial_iteration(5))
print("the factioral in recursion",factorial_recursion(5))
```
